```python
import matplotlib.pyplot as plt

from torch.optim.lr_scheduler import ExponentialLR, LinearLR
from torch import nn
from torch.optim import Adam
```

```python
optimizer = Adam(nn.Linear(1,1).parameters(), lr=1e-3)
scheduler = ExponentialLR(
    optimizer=optimizer,
    gamma = 1-1.75e-3
)

lr = [scheduler.get_last_lr()[-1]]
for _ in range(2000):
    scheduler.step()
    lr.append(scheduler.get_last_lr()[-1])


optimizer = Adam(nn.Linear(1,1).parameters(), lr=1e-3)
scheduler = ExponentialLR(
    optimizer=optimizer,
    gamma = 1-1e-3
)
# scheduler_linear = LinearLR(
#     optimizer,
#     start_factor=1.0,
#     end_factor=0.02,
#     total_iters=1500,
# )
lr_2 = [scheduler.get_last_lr()[-1]]
for _ in range(2000):
    scheduler.step()
    lr_2.append(scheduler.get_last_lr()[-1])

print(lr[800])
print(lr[1000])
print(lr[-1])
fig = plt.figure()
plt.plot(lr)
plt.plot(lr_2, label='2')
plt.legend()
plt.show()

fig = plt.figure()
plt.plot(lr)
plt.plot(lr_2, label='2')
plt.yscale('log')
plt.legend()
plt.show()
```
